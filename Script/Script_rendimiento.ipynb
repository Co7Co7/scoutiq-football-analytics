```python
import pandas as pd
import io
import os

ruta_html = r"c:\Users\PAOLA\Desktop\Proyectos personales\Rendimiento_vs_salario\Rendimiento\2022-2023 Serie A Playing Time _ FBref.html"
ruta_csv = r"c:\Users\PAOLA\Desktop\Proyectos personales\Rendimiento_vs_salario\Rendimiento\fbref_seriea_22_23.csv"

try:
    with open(ruta_html, 'r', encoding='utf-8') as f:
        html_content = f.read()

    # TRUCO: A veces FBRef esconde la tabla en comentarios HTML <!-- ... -->
    # Vamos a limpiar los comentarios para exponer la tabla oculta
    limpio = html_content.replace('<!--', '').replace('-->', '')
    all_dfs = pd.read_html(io.StringIO(limpio))

    print(f"Se encontraron {len(all_dfs)} tablas en total.")

    df_final = None
    for i, df in enumerate(all_dfs):
        # Aplanamos las columnas por si tienen niveles múltiples
        columnas = [str(col).lower() for col in df.columns.get_level_values(-1)]
        
        # Imprimimos las primeras columnas de cada tabla para que veas qué hay
        print(f"Tabla {i} columnas: {columnas[:5]}")

        # Buscamos 'player' o 'joueur' (por si se guardó en francés)
        if any(keyword in columnas for keyword in ['player', 'joueur', 'rk']):
            print(f"--> ¡Candidata encontrada en la Tabla {i}!")
            df_final = df
            break

    if df_final is not None:
        if isinstance(df_final.columns, pd.MultiIndex):
            df_final.columns = df_final.columns.droplevel(0)
        
        df_final = df_final[df_final['Player'] != 'Player']
        df_final.to_csv(ruta_csv, index=False, encoding='utf-8-sig')
        print(f"\n¡ÉXITO! Generado con {len(df_final)} registros.")
    else:
        print("\nERROR: Ninguna tabla contiene datos de jugadores. Revisa si el archivo HTML tiene contenido visible al abrirlo.")

except Exception as e:
    print(f"Error técnico: {e}")
```
